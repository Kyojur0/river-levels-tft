# River results

Load the bounded rolling-origin result JSON and compare models. Missing or unavailable models stay explicit.

```python
from pathlib import Path
import json
import pandas as pd

result = json.loads(Path('../results/river_baselines.json').read_text())
rows = []
for model, details in result['baselines'].items():
    metrics = details.get('metrics', {})
    rows.append({'model': model, 'status': details.get('status'), **metrics})
pd.DataFrame(rows).sort_values('p50_pinball', na_position='last')
```
